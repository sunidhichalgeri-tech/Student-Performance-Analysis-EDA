## 4. Correlation Analysis

Correlation analysis was performed to identify numerical factors
associated with students' final grades (G3).

### Key Findings

- G2 has the strongest positive correlation with G3 (0.905).
- G1 also has a strong positive correlation with G3 (0.801).
- Previous failures show a moderate negative correlation with G3 (-0.360).
- Mother's education has a weak positive correlation with G3 (0.217).
- Study time has a very weak positive correlation with G3 (0.098).
- Absences show almost no linear correlation with G3 (0.034).

These correlations indicate that previous academic performance and
past failures are more strongly associated with final grades than
several demographic and lifestyle factors.

```python
def performance_category(grade):
    if grade < 10:
        return "Low"
    elif grade < 15:
        return "Medium"
    else:
        return "High"

df["performance"] = df["G3"].apply(performance_category)
```

# 5. Key Insights

Based on the exploratory analysis, the following insights were identified:

1. Previous academic performance is strongly associated with final
   performance. G2 has the strongest correlation with G3 (0.905),
   followed by G1 (0.801).

2. Students with a greater number of previous failures tend to have
   lower final grades, with a correlation of -0.360.

3. Study time has only a weak positive correlation with final grade
   (0.098), suggesting that study duration alone does not strongly
   explain final academic performance.

4. Internet access is associated with a higher average final grade
   in this dataset.

5. Students receiving additional school support have a lower average
   grade than students who do not receive such support. This should
   not be interpreted as support causing lower performance, since
   students who are struggling may be more likely to receive support.

6. Absences show very little linear correlation with final grade
   (0.034).

7. Parental education shows a weak positive relationship with final
   academic performance.

8. Correlation analysis identifies associations between variables,
   but correlation does not imply causation.

## 6. Conclusion

This exploratory data analysis examined student academic performance
using demographic, academic, social, and school-related variables.

The analysis found that previous academic grades were the strongest
numerical factors associated with final performance. Past failures
showed a negative relationship with final grades, while factors such
as study time and absences showed relatively weak linear relationships.

Categorical analysis also revealed differences in average performance
based on internet access, school support, extracurricular activities,
paid classes, and higher education plans.

Overall, the analysis demonstrates how exploratory data analysis can
be used to identify patterns, relationships, and potential influencing
factors within student performance data.

These findings represent associations within the dataset and should
not be interpreted as proof of causal relationships.
